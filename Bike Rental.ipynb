```python
cols = ["yr","mnth","temp","atemp","hum","windspeed","casual","registered"]
```

```python
df[cols] = df[cols].replace("?", None)
df[cols] = df[cols].apply(lambda x: x.str.replace(",", "").str.strip())
df[cols] = df[cols].apply(pd.to_numeric)
```

```python
import numpy as np
for col in df.select_dtypes(include='object'):
    df[col] = df[col].str.replace(r'[^A-Za-z0-9 ]', '', regex=True)
    df[col] = df[col].replace('', np.nan)
```

```python
df = df.apply(lambda x: x.fillna(x.mode()[0]))

```

```python
df1 = df.iloc[:,[0,4,5,7,10,11,12,13,14,15,16]]
```

```python
df1
```

```python
import numpy as np
num_cols = df1
for col in num_cols: 
    Q1 = np.percentile(df1[col],25) 
    Q3 = np.percentile(df1[col],75) 
    print("Q1:",Q1) 
    print("Q3:",Q3) 
    IQR = Q3-Q1 
    UW = Q3+(1.5*IQR) 
    LW = Q1-(1.5*IQR) 
    print(f"Upper whisker length of {col}:",UW) 
    print(f"Lower whisker length of {col}:",LW) 
    print(f"Total no. of outliers in {col}:", len(df1[(df1[col] < LW) | (df1[col] > UW)]))
    df1.loc[df1[col] > UW, col] = UW
    df1.loc[df1[col] < LW, col] = LW
```

```python
X = df_final.drop('cnt', axis=1)   # features
Y = df_final['cnt']                # target
```

```python
X.shape
```

```python
# Data partition
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y, test_size=0.3)
print(X_train.shape,X_test.shape,Y_train.shape,Y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
LR =  LinearRegression()
LR.fit(X_train, Y_train)
Y_pred_train = LR.predict(X_train)
Y_pred_test = LR.predict(X_test)

from sklearn.metrics import mean_squared_error, r2_score,mean_absolute_error
import numpy as np

mse_train = mean_squared_error(Y_train, Y_pred_train)
rmse_train = np.sqrt(mse_train)
mae_train = mean_absolute_error(Y_train, Y_pred_train)
r2_train = r2_score(Y_train, Y_pred_train)

mse_test = mean_squared_error(Y_test, Y_pred_test)
rmse_test = np.sqrt(mse_test)
mae_test = mean_absolute_error(Y_train, Y_pred_train)
r2_test = r2_score(Y_test, Y_pred_test)


print("Mean Squared train Error:", np.round(mse_train, 2))
print("Root Mean Squared train Error:", np.round(rmse_train, 2))
print(" training R-squared:", np.round(r2_train, 2))
print("Mean absolute train error:",np.round(mae_train, 2))

print("Mean Squared test Error:", np.round(mse_test, 2))
print("Root Mean Squared test Error:", np.round(rmse_test, 2))
print(" testing R-squared:", np.round(r2_test, 2)) 
print("Mean absolute test error:",np.round(mae_test, 2))

```

```python
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import GridSearchCV

param_grid = {
    'n_estimators': [100, 200],
    'learning_rate': [0.05, 0.1],
    'max_depth': [3, 5]
}

gbr = GradientBoostingRegressor(random_state=42)

grid = GridSearchCV(
    estimator=gbr,
    param_grid=param_grid,
    cv=5,                   
    scoring='r2'
)

grid.fit(X_train, Y_train)
print(grid.best_params_)
```

```python
from sklearn.model_selection import RandomizedSearchCV

param_dist = {
    'n_estimators': [50, 100, 200, 300],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10]
}

random_search = RandomizedSearchCV(
    rf,
    param_dist,
    n_iter=10,
    cv=5,
    scoring='r2',
    random_state=42
)

random_search.fit(X_train, Y_train)

print(random_search.best_params_)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

models = {
    "Decision Tree": DT,
    "Random Forest": rf,
    "Gradient Boosting": gbr,
    "Linear Regression": LR,
    "Support Vector Regressor": svr
}

results = []

for name, model in models.items():
    model.fit(X_train, Y_train)
    Y_pred = model.predict(X_test)
    
    results.append({
        "Model": name,
        "MAE": round(mean_absolute_error(Y_test, Y_pred), 2),
        "RMSE": round(np.sqrt(mean_squared_error(Y_test, Y_pred)), 2),
        "R2": round(r2_score(Y_test, Y_pred), 2)
    })

print(results)
```
